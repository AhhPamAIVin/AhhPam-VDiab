```python
from __future__ import annotations

from functools import reduce
from pathlib import Path
import warnings

import numpy as np
import pandas as pd


# ============================================================
# 1. CONFIGURATION
# ============================================================

DATA_DIR = Path(r"C:\Users\ADMIN\Downloads\data\NHANES")
OUTPUT_DIR = DATA_DIR / "processed"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# NHANES 2017-March 2020 pre-pandemic files expected by this script.
EXPECTED_FILES = [
    "P_DEMO", "P_BMX", "P_BPXO", "P_GHB", "P_GLU",
    "P_HDL", "P_TCHOL", "P_TRIGLY", "P_DIQ", "P_BPQ",
    "P_MCQ", "P_PAQ", "P_RXQ_RX", "P_SLQ", "P_SMQ",
    "P_ALQ", "P_WHQ", "P_BIOPRO",
]

# Exact variables selected from the uploaded data dictionary.
COLUMN_MAPS: dict[str, dict[str, str]] = {
    "P_DEMO": {
        "SEQN": "patient_id",
        "RIDAGEYR": "age_years",
        "RIAGENDR": "sex_code",
        "RIDRETH3": "race_ethnicity_code",
        "INDFMPIR": "income_poverty_ratio",
    },
    "P_BMX": {
        "SEQN": "patient_id",
        "BMXWT": "weight_kg",
        "BMXHT": "height_cm",
        "BMXBMI": "bmi",
        "BMXWAIST": "waist_cm",
    },
    "P_GHB": {
        "SEQN": "patient_id",
        "LBXGH": "hba1c_pct",
    },
    "P_GLU": {
        "SEQN": "patient_id",
        "LBXGLU": "fasting_glucose_mg_dl",
    },
    "P_HDL": {
        "SEQN": "patient_id",
        "LBDHDD": "hdl_mg_dl",
    },
    "P_TCHOL": {
        "SEQN": "patient_id",
        "LBXTC": "total_cholesterol_mg_dl",
    },
    "P_TRIGLY": {
        "SEQN": "patient_id",
        "LBXTR": "triglycerides_mg_dl",
        "LBDLDL": "ldl_mg_dl",
    },
    "P_BPXO": {
        "SEQN": "patient_id",
        "BPXOSY1": "sbp_1",
        "BPXODI1": "dbp_1",
        "BPXOSY2": "sbp_2",
        "BPXODI2": "dbp_2",
        "BPXOSY3": "sbp_3",
        "BPXODI3": "dbp_3",
        "BPXOPLS1": "pulse_1",
        "BPXOPLS2": "pulse_2",
        "BPXOPLS3": "pulse_3",
    },
    "P_DIQ": {
        "SEQN": "patient_id",
        "DIQ010": "doctor_diagnosed_diabetes_code",
        "DIQ160": "doctor_diagnosed_prediabetes_code",
        "DIQ050": "taking_insulin_code",
        "DIQ070": "taking_diabetes_pills_code",
    },
    "P_BPQ": {
        "SEQN": "patient_id",
        "BPQ020": "doctor_diagnosed_hypertension_code",
        "BPQ040A": "advised_bp_medication_code",
        "BPQ050A": "taking_bp_medication_code",
        "BPQ080": "doctor_diagnosed_high_cholesterol_code",
        "BPQ100D": "taking_cholesterol_medication_code",
    },
    "P_MCQ": {
        "SEQN": "patient_id",
        "MCQ160B": "congestive_heart_failure_code",
        "MCQ160C": "coronary_heart_disease_code",
        "MCQ160D": "angina_code",
        "MCQ160E": "heart_attack_code",
        "MCQ160F": "stroke_code",
        "MCQ160M": "thyroid_problem_code",
        "MCQ160L": "liver_condition_code",
        "MCQ300C": "family_history_diabetes_code",
    },
    "P_PAQ": {
        "SEQN": "patient_id",
        "PAQ605": "vigorous_work_activity_code",
        "PAQ610": "vigorous_work_days_week",
        "PAD615": "vigorous_work_minutes_day",
        "PAQ620": "moderate_work_activity_code",
        "PAQ625": "moderate_work_days_week",
        "PAD630": "moderate_work_minutes_day",
        "PAQ635": "walk_bicycle_transport_code",
        "PAQ650": "vigorous_recreation_code",
        "PAQ665": "moderate_recreation_code",
        "PAD680": "sedentary_minutes_day",
    },
    "P_SLQ": {
        "SEQN": "patient_id",
        "SLD012": "sleep_hours_weekday",
        "SLD013": "sleep_hours_weekend",
        "SLQ050": "doctor_diagnosed_sleep_disorder_code",
    },
    "P_SMQ": {
        "SEQN": "patient_id",
        "SMQ020": "smoked_100_cigarettes_code",
        "SMQ040": "current_smoking_frequency_code",
    },
    "P_ALQ": {
        "SEQN": "patient_id",
        "ALQ111": "alcohol_ever_code",
        "ALQ121": "alcohol_frequency_past_year",
        "ALQ130": "avg_drinks_per_drinking_day",
    },
    "P_WHQ": {
        "SEQN": "patient_id",
        "WHD020": "self_reported_current_weight_lb",
        "WHD050": "self_reported_weight_year_ago_lb",
        "WHD110": "self_reported_weight_10_years_ago_lb",
        "WHD140": "self_reported_max_weight_lb",
    },
    "P_BIOPRO": {
        "SEQN": "patient_id",
        "LBXSCR": "creatinine_mg_dl",
        "LBXSATSI": "alt_u_l",
        "LBXSASSI": "ast_u_l",
        "LBXSGTSI": "ggt_u_l",
        "LBXSBU": "bun_mg_dl",
    },
}


# ============================================================
# 2. HELPERS
# ============================================================

def find_xpt_file(dataset_name: str) -> Path | None:
    """Find an XPT file case-insensitively."""
    candidates = list(DATA_DIR.glob(f"{dataset_name}.XPT"))
    candidates += list(DATA_DIR.glob(f"{dataset_name}.xpt"))
    return candidates[0] if candidates else None


def decode_bytes_columns(df: pd.DataFrame) -> pd.DataFrame:
    """Decode byte strings created by read_sas."""
    for col in df.select_dtypes(include=["object"]).columns:
        df[col] = df[col].map(
            lambda x: x.decode("utf-8", errors="replace").strip()
            if isinstance(x, (bytes, bytearray))
            else x
        )
    return df


def read_xpt(dataset_name: str) -> pd.DataFrame | None:
    file_path = find_xpt_file(dataset_name)
    if file_path is None:
        warnings.warn(f"Không tìm thấy file {dataset_name}.XPT")
        return None

    print(f"Reading {file_path.name} ...")
    df = pd.read_sas(file_path, format="xport")
    df.columns = [str(c).upper() for c in df.columns]
    return decode_bytes_columns(df)


def select_and_rename(
    df: pd.DataFrame | None,
    dataset_name: str,
    column_map: dict[str, str],
) -> pd.DataFrame | None:
    """Select only columns that really exist; report missing expected columns."""
    if df is None:
        return None

    available = [col for col in column_map if col in df.columns]
    missing = [col for col in column_map if col not in df.columns]

    if missing:
        warnings.warn(
            f"{dataset_name}: thiếu biến dự kiến {missing}. "
            "Script vẫn tiếp tục với các biến hiện có."
        )

    if "SEQN" not in available:
        warnings.warn(f"{dataset_name}: không có SEQN nên bỏ qua file.")
        return None

    result = df[available].copy().rename(columns=column_map)
    result["patient_id"] = pd.to_numeric(
        result["patient_id"], errors="coerce"
    ).astype("Int64")
    return result


def normalize_nhanes_yes_no(series: pd.Series) -> pd.Series:
    """
    Standard NHANES questionnaire convention:
      1 = Yes, 2 = No; 7/9 and other codes are treated as missing.
    """
    numeric = pd.to_numeric(series, errors="coerce")
    return numeric.map({1: 1, 2: 0}).astype("Int64")


def first_non_null(values: pd.Series):
    values = values.dropna()
    return values.iloc[0] if not values.empty else np.nan


def aggregate_medications(rx_df: pd.DataFrame | None) -> pd.DataFrame | None:
    """
    P_RXQ_RX is one-to-many: one participant can have many drug rows.
    Aggregate names and indication descriptions before merging.
    """
    if rx_df is None or "SEQN" not in rx_df.columns:
        return None

    keep = [
        col for col in
        ["SEQN", "RXDUSE", "RXDDRUG", "RXDDRGID", "RXDRSC1", "RXDRSD1", "RXDDAYS"]
        if col in rx_df.columns
    ]
    rx = rx_df[keep].copy()
    rx = decode_bytes_columns(rx)
    rx["SEQN"] = pd.to_numeric(rx["SEQN"], errors="coerce").astype("Int64")

    def join_unique(series: pd.Series) -> str | float:
        clean = (
            series.dropna()
            .astype(str)
            .str.strip()
            .replace({"": np.nan, "nan": np.nan})
            .dropna()
        )
        values = sorted(set(clean))
        return " | ".join(values) if values else np.nan

    agg_rules = {}
    if "RXDDRUG" in rx.columns:
        agg_rules["RXDDRUG"] = join_unique
    if "RXDDRGID" in rx.columns:
        agg_rules["RXDDRGID"] = join_unique
    if "RXDRSC1" in rx.columns:
        agg_rules["RXDRSC1"] = join_unique
    if "RXDRSD1" in rx.columns:
        agg_rules["RXDRSD1"] = join_unique
    if "RXDDAYS" in rx.columns:
        agg_rules["RXDDAYS"] = "max"
    if "RXDUSE" in rx.columns:
        agg_rules["RXDUSE"] = first_non_null

    grouped = rx.groupby("SEQN", as_index=False).agg(agg_rules)

    grouped = grouped.rename(columns={
        "SEQN": "patient_id",
        "RXDDRUG": "medication_names",
        "RXDDRGID": "medication_drug_ids",
        "RXDRSC1": "medication_indication_codes",
        "RXDRSD1": "medication_indication_descriptions",
        "RXDDAYS": "maximum_medication_days",
        "RXDUSE": "prescription_medication_use_code",
    })
    return grouped


# ============================================================
# 3. LOAD AND CLEAN
# ============================================================

raw: dict[str, pd.DataFrame] = {}
clean_tables: list[pd.DataFrame] = []

for dataset_name in EXPECTED_FILES:
    df = read_xpt(dataset_name)
    if df is not None:
        raw[dataset_name] = df

    if dataset_name in COLUMN_MAPS:
        cleaned = select_and_rename(df, dataset_name, COLUMN_MAPS[dataset_name])
        if cleaned is not None:
            # Defensive check for one-row-per-person datasets.
            duplicates = cleaned["patient_id"].duplicated().sum()
            if duplicates:
                warnings.warn(
                    f"{dataset_name}: có {duplicates} SEQN trùng. "
                    "Giữ dòng đầu tiên trước khi merge."
                )
                cleaned = cleaned.drop_duplicates("patient_id", keep="first")
            clean_tables.append(cleaned)

# Aggregate prescription medications separately.
rx_agg = aggregate_medications(raw.get("P_RXQ_RX"))
if rx_agg is not None:
    clean_tables.append(rx_agg)

if not clean_tables:
    raise RuntimeError("Không đọc được bảng hợp lệ nào. Kiểm tra lại DATA_DIR.")

# Use demographics as the master population whenever available.
demo_table = next(
    (table for table in clean_tables if "age_years" in table.columns),
    clean_tables[0],
)
other_tables = [t for t in clean_tables if t is not demo_table]

clinical = demo_table.copy()
for table in other_tables:
    clinical = clinical.merge(
        table,
        on="patient_id",
        how="left",
        validate="one_to_one",
    )


# ============================================================
# 4. DERIVED VARIABLES
# ============================================================

# Average all available BP/pulse readings row-wise.
sbp_cols = [c for c in ["sbp_1", "sbp_2", "sbp_3"] if c in clinical.columns]
dbp_cols = [c for c in ["dbp_1", "dbp_2", "dbp_3"] if c in clinical.columns]
pulse_cols = [c for c in ["pulse_1", "pulse_2", "pulse_3"] if c in clinical.columns]

if sbp_cols:
    clinical["sbp_mean"] = clinical[sbp_cols].mean(axis=1, skipna=True)
if dbp_cols:
    clinical["dbp_mean"] = clinical[dbp_cols].mean(axis=1, skipna=True)
if pulse_cols:
    clinical["pulse_mean"] = clinical[pulse_cols].mean(axis=1, skipna=True)

# Decode selected yes/no fields to nullable 0/1 flags.
yes_no_fields = [
    "doctor_diagnosed_diabetes_code",
    "doctor_diagnosed_prediabetes_code",
    "taking_insulin_code",
    "taking_diabetes_pills_code",
    "doctor_diagnosed_hypertension_code",
    "taking_bp_medication_code",
    "doctor_diagnosed_high_cholesterol_code",
    "taking_cholesterol_medication_code",
    "congestive_heart_failure_code",
    "coronary_heart_disease_code",
    "angina_code",
    "heart_attack_code",
    "stroke_code",
    "thyroid_problem_code",
    "liver_condition_code",
    "family_history_diabetes_code",
    "vigorous_work_activity_code",
    "moderate_work_activity_code",
    "walk_bicycle_transport_code",
    "vigorous_recreation_code",
    "moderate_recreation_code",
    "doctor_diagnosed_sleep_disorder_code",
    "smoked_100_cigarettes_code",
    "alcohol_ever_code",
]
for col in yes_no_fields:
    if col in clinical.columns:
        flag_name = col.removesuffix("_code") + "_flag"
        clinical[flag_name] = normalize_nhanes_yes_no(clinical[col])

# Helpful sex labels while preserving raw code.
if "sex_code" in clinical.columns:
    clinical["sex"] = (
        pd.to_numeric(clinical["sex_code"], errors="coerce")
        .map({1: "Male", 2: "Female"})
    )

# Diabetes/prediabetes evidence.
hba1c = pd.to_numeric(
    clinical.get("hba1c_pct", pd.Series(index=clinical.index, dtype=float)),
    errors="coerce",
)
fpg = pd.to_numeric(
    clinical.get(
        "fasting_glucose_mg_dl",
        pd.Series(index=clinical.index, dtype=float),
    ),
    errors="coerce",
)

diagnosed_diabetes = (
    clinical.get(
        "doctor_diagnosed_diabetes_flag",
        pd.Series(pd.NA, index=clinical.index, dtype="Int64"),
    ) == 1
)
diagnosed_prediabetes = (
    clinical.get(
        "doctor_diagnosed_prediabetes_flag",
        pd.Series(pd.NA, index=clinical.index, dtype="Int64"),
    ) == 1
)

lab_diabetes = (hba1c >= 6.5) | (fpg >= 126)
lab_prediabetes = (
    hba1c.between(5.7, 6.4, inclusive="both")
    | fpg.between(100, 125, inclusive="both")
)

has_hba1c = hba1c.notna()
has_fpg = fpg.notna()
has_any_lab = has_hba1c | has_fpg

# Conservative healthy rule: every available glycemic measurement must be normal.
lab_normal = (
    ((~has_hba1c) | (hba1c < 5.7))
    & ((~has_fpg) | (fpg < 100))
    & has_any_lab
)

clinical["diabetes_stage"] = pd.Series(
    "Unclassified", index=clinical.index, dtype="string"
)

# Priority: diabetes overrides prediabetes.
clinical.loc[lab_normal & ~diagnosed_prediabetes & ~diagnosed_diabetes,
             "diabetes_stage"] = "Healthy"
clinical.loc[(lab_prediabetes | diagnosed_prediabetes) & ~diagnosed_diabetes,
             "diabetes_stage"] = "Prediabetes"
clinical.loc[lab_diabetes | diagnosed_diabetes,
             "diabetes_stage"] = "Diabetes"

# Evidence columns make the classification auditable.
clinical["diabetes_by_lab_flag"] = lab_diabetes.astype("Int64")
clinical["prediabetes_by_lab_flag"] = lab_prediabetes.astype("Int64")
clinical["normal_glycemia_by_lab_flag"] = lab_normal.astype("Int64")

# Common derived risk flags for modelling, not medical diagnosis.
if "bmi" in clinical.columns:
    bmi = pd.to_numeric(clinical["bmi"], errors="coerce")
    clinical["bmi_category"] = pd.cut(
        bmi,
        bins=[-np.inf, 18.5, 25, 30, np.inf],
        labels=["Underweight", "Normal", "Overweight", "Obesity"],
        right=False,
    )

if "sbp_mean" in clinical.columns and "dbp_mean" in clinical.columns:
    clinical["elevated_bp_screen_flag"] = (
        (clinical["sbp_mean"] >= 130) | (clinical["dbp_mean"] >= 80)
    ).astype("Int64")

clinical["source_dataset"] = "NHANES_2017_2020"
clinical["source_patient_id"] = clinical["patient_id"].astype("string")


# ============================================================
# 5. DATA QUALITY REPORT
# ============================================================

quality = pd.DataFrame({
    "column": clinical.columns,
    "dtype": [str(clinical[c].dtype) for c in clinical.columns],
    "missing_count": [int(clinical[c].isna().sum()) for c in clinical.columns],
    "missing_pct": [
        round(float(clinical[c].isna().mean() * 100), 2)
        for c in clinical.columns
    ],
    "unique_count": [
        int(clinical[c].nunique(dropna=True))
        for c in clinical.columns
    ],
})
quality = quality.sort_values(
    ["missing_pct", "column"], ascending=[False, True]
)

stage_counts = (
    clinical["diabetes_stage"]
    .value_counts(dropna=False)
    .rename_axis("diabetes_stage")
    .reset_index(name="participant_count")
)

# Prediabetes cohort for the user's project.
prediabetes = clinical.loc[
    clinical["diabetes_stage"] == "Prediabetes"
].copy()


# ============================================================
# 6. EXPORT
# ============================================================

clinical_path = OUTPUT_DIR / "nhanes_clinical_master.csv"
prediabetes_path = OUTPUT_DIR / "nhanes_prediabetes_cohort.csv"
quality_path = OUTPUT_DIR / "nhanes_data_quality.csv"
stage_path = OUTPUT_DIR / "nhanes_stage_counts.csv"

clinical.to_csv(clinical_path, index=False, encoding="utf-8-sig")
prediabetes.to_csv(prediabetes_path, index=False, encoding="utf-8-sig")
quality.to_csv(quality_path, index=False, encoding="utf-8-sig")
stage_counts.to_csv(stage_path, index=False, encoding="utf-8-sig")

print("\n================ COMPLETED ================")
print(f"Clinical master:      {clinical_path}")
print(f"Prediabetes cohort:   {prediabetes_path}")
print(f"Data quality report:  {quality_path}")
print(f"Stage counts:         {stage_path}")
print("\nDiabetes stage distribution:")
print(stage_counts.to_string(index=False))
print("\nMaster shape:", clinical.shape)
print("Prediabetes shape:", prediabetes.shape)
```
